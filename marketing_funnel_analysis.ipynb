```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('bank-full.csv', sep=';')
df.head()
```

```python
conversion = df['y'].value_counts()
conversion
```

```python
conversion.plot(kind='bar')

plt.title('Lead Conversion Results')
plt.xlabel('Conversion')
plt.ylabel('Number of Customers')

plt.show()
```

```python
conversion_rate = (5289 / 45211) * 100
print("Conversion Rate:", round(conversion_rate, 2), "%")
```

```python
channel_conversion = pd.crosstab(df['contact'], df['y'])

channel_conversion.plot(kind='bar')

plt.title('Conversion by Contact Type')
plt.xlabel('Contact Type')
plt.ylabel('Number of Customers')

plt.show()
```
